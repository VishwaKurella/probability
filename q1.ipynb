# Problem Statement

A drawer containes red socks and black socks. When two scoks are drwan at random, the probability that both are red is $1\over2$. 
1. How small can the number of socks in the drawer be?
2. How small if the number of black socks is even?

# Solution

Let \(r\) be the number of red socks and \(b\) the number of black socks.  
Total socks:

$$
T=r+b
$$

The probability of drawing two red socks without replacement is

$$
P(RR)=\frac{r}{T}\cdot\frac{r-1}{T-1}
$$

Given

$$
P(RR)=\frac12
$$

we get

$$
\frac{r(r-1)}{T(T-1)}=\frac12
$$

so

$$
2r(r-1)=T(T-1)
$$

Expanding:

$$
2r^2-2r=T^2-T
$$

Complete the square:

$$
(2T-1)^2-2(2r-1)^2=-1
$$

Let

$$
x=2T-1,\qquad y=2r-1
$$

Then

$$
x^2-2y^2=-1
$$

which is a Pell equation.

The first few positive solutions are

$$
(1,1),(7,5),(41,29),\dots
$$

Using

$$
T=\frac{x+1}{2},\qquad r=\frac{y+1}{2},\qquad b=T-r
$$

we obtain:

| \(x\) | \(y\) | \(r\) | \(b\) |
|---|---|---|---|
| 1 | 1 | 1 | 0 |
| 7 | 5 | 3 | 1 |
| 41 | 29 | 15 | 6 |

Hence:

- Smallest non-trivial solution:

$$
\boxed{r=3,\ b=1}
$$

- Smallest solution with even black socks:

$$
\boxed{r=15,\ b=6}
$$

Checks:

$$
\frac34\cdot\frac23=\frac12,
\qquad
\frac{15}{21}\cdot\frac{14}{20}=\frac12
$$


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 10000)
y = np.sqrt((x*x + 1) / 2)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.show()
```

```python
from scipy.optimize import fsolve

def eq(v):
    x, y = v
    return [
        x*x - 2*y*y + 1,
        y - np.sqrt((x*x + 1)/2)
    ]

guesses = [
    (1,1),
    (7,5),
    (40,30),
    (240,170)
]

solutions = set()

for g in guesses:
    sol = fsolve(eq, g)

    x = round(sol[0])
    y = round(sol[1])

    if x*x - 2*y*y == -1:
        solutions.add((x,y))

print(sorted(solutions))
```
